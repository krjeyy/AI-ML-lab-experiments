```python
import heapq

# Define a Graph structure with edge weights
class Graph:
    def __init__(self):
        self.edges = {}
        self.heuristics = {}

    def add_edge(self, u, v, cost):
        if u not in self.edges:
            self.edges[u] = []
        if v not in self.edges:
            self.edges[v] = []
        self.edges[u].append((v, cost))
        # Assuming an undirected graph for simplicity
        self.edges[v].append((u, cost))

    def set_heuristic(self, node, value):
        self.heuristics[node] = value

    def get_neighbors(self, node):
        return self.edges.get(node, [])

    def get_heuristic(self, node):
        return self.heuristics.get(node, 0)
```

```python
def uniform_cost_search(graph, start, goal):
    # Priority Queue stores tuples of (cumulative_cost, current_node, path_taken)
    priority_queue = [(0, start, [start])]
    visited = set()

    while priority_queue:
        cost, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            return path, cost

        for neighbor, edge_cost in graph.get_neighbors(current):
            if neighbor not in visited:
                heapq.heappush(priority_queue, (cost + edge_cost, neighbor, path + [neighbor]))

    return None, float('inf')
```

```python
def a_star_search(graph, start, goal):
    # Priority Queue stores (f_score, g_score, current_node, path)
    # f_score = g_score + heuristic
    start_heuristic = graph.get_heuristic(start)
    priority_queue = [(start_heuristic, 0, start, [start])]
    visited = set()

    while priority_queue:
        f_score, g_score, current, path = heapq.heappop(priority_queue)

        if current in visited:
            continue

        visited.add(current)

        if current == goal:
            return path, g_score

        for neighbor, edge_cost in graph.get_neighbors(current):
            if neighbor not in visited:
                new_g_score = g_score + edge_cost
                new_f_score = new_g_score + graph.get_heuristic(neighbor)
                heapq.heappush(priority_queue, (new_f_score, new_g_score, neighbor, path + [neighbor]))

    return None, float('inf')
```

## Testing both Algorithms

```python
# Create sample graph
g = Graph()
g.add_edge('A', 'B', 1)
g.add_edge('A', 'C', 4)
g.add_edge('B', 'C', 2)
g.add_edge('B', 'D', 5)
g.add_edge('C', 'D', 1)

# Heuristic values for nodes targeting goal 'D'
g.set_heuristic('A', 3)
g.set_heuristic('B', 2)
g.set_heuristic('C', 1)
g.set_heuristic('D', 0)

# Run UCS
ucs_path, ucs_cost = uniform_cost_search(g, 'A', 'D')
print(f"UCS Path: {ucs_path} with Cost: {ucs_cost}")

# Run A*
ast_path, ast_cost = a_star_search(g, 'A', 'D')
print(f"A* Path: {ast_path} with Cost: {ast_cost}")
```
